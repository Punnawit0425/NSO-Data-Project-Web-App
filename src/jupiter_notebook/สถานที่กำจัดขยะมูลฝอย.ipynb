```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/location.csv')
df
```

```python
df.isnull().sum()
df.info()
```

```python
for col in df.columns[1:4]:
    df[col] = df[col].astype('category')

df.info()
```

```python
df['สถานที่กำจัดขยะ(แห่ง)'] = df['สถานที่กำจัดขยะ(แห่ง)'].fillna(df['สถานที่กำจัดขยะ(แห่ง)'].median())
df.info()
```

```python
df.to_csv('../clean_data/clean_location.csv')
```
