```python
for col in df.columns[1:5]:
    df[col] = df[col].astype('category')
df['ขยะรีไซเคิลลงถัง(ตัน)'] = df['ขยะรีไซเคิลลงถัง(ตัน)'].str.replace(',', '', regex=False).astype(float)
df.info()
```

```python
# หาค่ากลาง (Median) ของแต่ละ อปท.
df['median_val'] = df.groupby('อปท')['ขยะรีไซเคิลลงถัง(ตัน)'].transform('median')

# เงื่อนไข: ถ้าเกิน 10 เท่าของค่ากลาง และมากกว่า 500 ตัน ให้แก้
condition = (df['ขยะรีไซเคิลลงถัง(ตัน)'] > df['median_val'] * 10) & (df['ขยะรีไซเคิลลงถัง(ตัน)'] > 500)
df.loc[condition, 'ขยะรีไซเคิลลงถัง(ตัน)'] = df.loc[condition, 'median_val']

# เซฟไฟล์
df.drop(columns=['median_val'])
```

```python
df.to_csv('../clean_data/clean_recycle.csv')
```
